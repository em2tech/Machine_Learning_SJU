**Question 2: How did you determine which columns needed imputation, and what strategy did you choose?**

I determined which columns needed imputation by running df.. insull().sum() accross the full DataFrame, which quantifies missing values per column rather than relying on guesswork.

In this dataset, that check revealed encryption_used had 1,966 missing values, while the other columns showed no missing data. For the imputation strategy itself, I split approach by data type, since a single strategy doesn't make sense across both:

•	Numeric columns (e.g., session_duration, failed_logins) used median imputation, since the median is robust to outliers, unlike the mean, which can be pulled in one direction by extreme values.

•	Categorical columns (including encryption_used) used most-frequent imputation, filling missing entries with whichever category appeared most often in that column, since averaging doesn't make sense for text-based categories.


**Question 3: How do ColumnTransformer and Pipeline prevent data leakage?**

Pipeline and ColumnTransformer combine multiple preprocessing steps (like filling missing values, scaling, and encoding) into one object that runs everything in a single call, instead of applying each step manually. This helps in two ways:

**Consistency**:  every column goes through the same steps in the same order every time, so nothing gets applied out of order or skipped by accident.

**Preventing data leakage**: when a pipeline "learns" something from the data (like a median value or list of categories), that should only come from the training data, not the test data. If the test data influences these calculations, the model gets an unfair sneak peek at data it's supposed to be evaluated on later. A pipeline makes it easy to fit only on the training set, then apply those same rules to the test set; keeping the two properly separated.
